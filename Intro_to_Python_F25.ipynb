## 1. Loading Libraries and Data

Python by itself has standard tools. To perform specialized tasks, like data analysis, we import libraries (or additional "toolboxes").
- **Pandas (pd)**: The essential library for working with tabular data (i.e., tables, like in a spreadsheet). We use it to load, clean, and analyze our data.
- **Matplotlib (plt)**: The foundational library for creating static, publication-quality plots.
- **Seaborn (sns)**: A high-level library built on Matplotlib. It simplifies the creation of complex and attractive statistical visualizations.

**Import the Libraries**

Run the code cell below to import these libraries. We use the `as` keyword to give each one a short, standard alias.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**Load the Data**

Now, let's load our sample dataset. In this workshop, we're using a famous dataset (the Titanic passenger list) that comes built-in with the Seaborn library.

We will load it and store it in a variable named `df`. This is a common name for a Pandas "DataFrame," the object that holds our table.

```python
df = sns.load_dataset('titanic')
```

## 3. Cleaning Our Data
Remember in the previous step when we ran `df.describe()`? If you look at the `count` row, you'll see that `age` has a count of 714, while most other columns have 891. This tells us we have 177 rows where the passenger's age is unknown.

**Let's make a cleaning decision:** for this analysis, we will drop any rows that are missing an `age` value.

This ensures that when we analyze age, we're looking at a complete and consistent set of data.

- `subset=['age']` tells Pandas to only look for missing values in the age column.
- `inplace=True` modifies our df variable directly.

```python
df.dropna(subset=['age'], inplace=True)
```

Now, let's verify our change by running `df.describe()` one more time.

Look at the `count` row. You should now see that age, survived, pclass, and all other columns show the same count (714). This confirms our rows with missing ages have been successfully removed.

```python
df.describe()
```

Our data is now cleaner, and we're ready to start plotting.

## 4a. Making a Plot with Matplotlib

Numbers and tables are great, but a picture is worth a thousand words. Data visualization is where Python really shines.

Let's start with a **histogram** to see the distribution of passenger ages.

**Step 1: The Basic Plot**

 First, let's just create a basic histogram of the `age` column.

```python
plt.hist(df['age'])

plt.show()
```

**Step 2: The Polished Plot**

 That's a good start, but we can make it much clearer. Let's add:

- `bins=20` to see the distribution in more detail.
- `edgecolor='black'` to make the bars easier to see.
- Labels and a title to explain what the chart is.

```python
plt.hist(df['age'],bins=20)
plt.show()
```

```python
plt.hist(df['age'],bins=20,edgecolor='black')
plt.show()
```

```python
plt.hist(df['age'],bins=20,edgecolor='black')

plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Count')

plt.show()
```

## 4b. Making a Plot with Seaborn

While Matplotlib gives us full control, the **Seaborn** library (which we imported as `sns`) is excellent for creating common statistical plots in a single line.

Let's use Seaborn's `countplot` to see how many passengers were in each class.

```python
sns.countplot(data=df, x='class')
plt.show()
```

```python
sns.countplot(data=df, x='class')

plt.title('Total Passenger Count by Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count')

plt.show()
```

**Adding a Second Variable with `hue`**

That's useful, but what if we want to see the survival outcome within each class?

We can add the `hue` parameter to split the bars by a second categorical variable. This one line of code creates a very insightful chart.

```python
sns.countplot(data=df, x='class', hue='survived')

plt.title('Survival Count by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count')

plt.show()
```

This chart tells a much richer story, clearly showing the number of survivors and non-survivors in each class.
